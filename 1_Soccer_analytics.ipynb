# 1. Load the dataset and inspect it

```python
import pandas as pd
import numpy as np
from sympy import false

df = pd.read_csv("fifa_fbref_merged.csv")

print("Rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())

print("\nData types:")
print(df.dtypes)

print("\nBasic information:")
df.info()



```

# 2. Clean the column names first

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("/", "_per_", regex=False)
      .str.replace("%", "_pct", regex=False)
      .str.replace(r"[()]", "", regex=True)
)

print(df.columns.tolist())
```

```python
# Look at the most important identifying columns

important_cols = [
    "fifa_version",
    "season",
    "short_name",
    "long_name",
    "player",
    "club_name",
    "team",
    "league_name",
    "league",
    "player_positions",
    "pos",
    "age_fifa",
    "age_fbref",
    "playing_time_min"
]

print(df[important_cols].head(20))
```

```python
comparison_cols = [
    "short_name",
    "long_name",
    "player",
    "club_name",
    "team",
    "league_name",
    "league",
    "player_positions",
    "pos",
    "age_fifa",
    "age_fbref"
]

print(df[comparison_cols].head(20))
```

# 3. Look for missing data

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": (df.isnull().mean() * 100).round(2)
})

missing_summary = missing_summary.sort_values(
    "missing_percent",
    ascending=False
)

print(missing_summary.head(50))
```

# 4. Check duplicates

```python
print("Exact duplicate rows:", df.duplicated().sum())

duplicates = df[
    df.duplicated(
        subset=["season", "player", "team"],
        keep=False
    )
]

print("Player-season-team duplicate rows:", len(duplicates))

print(
    duplicates[
        ["season", "player", "team", "playing_time_min"]
    ].head(30)
)
```

# 5. Look at different positions

```python
print(df["pos"].value_counts(dropna=False))

print(df["player_positions"].value_counts(dropna=False).head(30))
```

# 7. Simplify attribute names further

```python

rename_map = {

    "per_90_minutes_gls": "goals_per90",

    "per_90_minutes_ast": "assists_per90",

    "per_90_minutes_xg": "xg_per90",

    "per_90_minutes_xag": "xag_per90",

    "per_90_minutes_npxg": "npxg_per90",

    "per_90_minutes_g+a": "goal_assist_per90",

    "playing_time_min": "minutes",

    "playing_time_90s": "full_90s",

    "playing_time_mp": "matches",

    "playing_time_starts": "starts",

    "total_cmp_pct": "pass_completion_pct"
}

df.rename(columns=rename_map, inplace=True)
```

# 8. Check data types and minute distribution

```python
print(
    df[
        [
            "season",
            "player",
            "team",
            "age_fbref",
            "minutes",
            "goals_per90",
            "assists_per90",
            "xg_per90"
        ]
    ].dtypes
)

print(df["minutes"].describe())
print(
    df["minutes"]
      .quantile([0, .10, .25, .50, .75, .90, .95, 1])
)
```

# 9. check number of season stored

```python
print(df["season"].value_counts().sort_index())
```

# 10. full data cleaning

```python
# ---------------------------------------------------
# DATA QUALITY ASSESSMENT
# ---------------------------------------------------

# 1. Dataset size
print("Dataset shape:", df.shape)


# 2. Missing values
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": (df.isnull().mean() * 100).round(2)
})

missing_summary = missing_summary.sort_values(
    "missing_percent",
    ascending=False
)

print("\nTop missing columns:")
print(missing_summary.head(30))


# 3. Exact duplicates
print("\nExact duplicate rows:")
print(df.duplicated().sum())


# 4. Player-season-team duplicates
player_duplicates = df.duplicated(
    subset=["season", "player", "team"],
    keep=False
)

print("\nPlayer-season-team duplicates:")
print(player_duplicates.sum())


# 5. Season distribution
print("\nSeasons:")
print(
    df["season"]
      .value_counts()
      .sort_index()
)


# 6. FBref position distribution
print("\nFBref Positions:")
print(
    df["pos"]
      .value_counts(dropna=False)
)


# 7. FIFA positions
print("\nFIFA Positions:")
print(
    df["player_positions"]
      .value_counts(dropna=False)
      .head(30)
)


# 8. Playing time
print("\nMinutes distribution:")
print(df["minutes"].describe())

print("\nMinutes quantiles:")
print(
    df["minutes"]
      .quantile([0, .10, .25, .50, .75, .90, .95, 1])
)

df.to_csv("soccer_players_clean.csv",index=false)
df_analysis = df.copy()
```
